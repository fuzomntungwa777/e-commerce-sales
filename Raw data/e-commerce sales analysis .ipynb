```python
#convert order_date data type to datetime
df['order_date'] = pd.to_datetime(df['order_date'], errors= 'coerce')
```

```python
#filling nulls
df['return_status'] = df['return_status'].fillna('Not returned')
df['return_reason'] = df['return_reason'].fillna('N/A')
df['campaign_name'] = df['return_reason'].fillna('None')
df['coupon_code'] = df['return_reason'].fillna('None')
```
